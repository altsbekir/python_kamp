```python
import requests
import json
```

```python
# GET İSTEĞİ

user_id = input("Enter id: ")
get_url = f"https://jsonplaceholder.typicode.com/todos/{user_id}"

get_response = requests.get(get_url)
print(get_response.json())
```

```python
# POST İSTEĞİ

post_url = "https://jsonplaceholder.typicode.com/todos"
todo_item = {"UserId" : 2,
             "title" : "bekir",
             "completed" : True}
post_response = requests.post(post_url, data=todo_item)
print(post_response.json())
```

```python
# optional header
headers = {"Content-Type" : "application/json"}
post_response = requests.post(post_url, json=todo_item, headers=headers)
print(post_response.json())
```

```python
post_response = requests.post(post_url, data=json.dumps(todo_item), headers=headers) # sözlüğü json'a çevirmek
print(post_response.json())
```

```python
# PUT İSTEĞİ

put_url = "https://jsonplaceholder.typicode.com/todos/15"
todo_item_15 = {"UserId" : 2,
                "title" : "put title",
                "completed": False}
put_response = requests.put(put_url, data=json.dumps(todo_item_15))
print(put_response.json())
```

```python
# PATCH İSTEĞİ

patch_url = "https://jsonplaceholder.typicode.com/todos/15"
todo_item_15 = {"title" : "patch title"}
patch_response = requests.patch(patch_url, data=json.dumps(todo_item_15))
print(patch_response.json())
```

```python
# DELETE İSTEĞİ

delete_url = "https://jsonplaceholder.typicode.com/todos/15"
delete_response = requests.delete(delete_url)
print(delete_response.status_code)
```
