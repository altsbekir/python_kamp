```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
age_list = [20, 25, 30, 35, 40, 45, 50, 55, 60]
weight_list = [70, 75, 73, 85, 89, 90, 75, 84, 90]
age_list_np = np.array(age_list)
weight_list_np = np.array(weight_list)
```

```python
plt.plot(age_list_np, weight_list_np, "c") # x = age, y = weight, "c" = cyan
plt.xlabel("age") # x eksenine age ismini verelim
plt.ylabel("weight") # y eksenine weight ismini verelim
plt.title("age - weight graph") # grafik ismi belirleme
```

```python
my_np1 = np.linspace(0, 15, 20)
my_np2 = my_np1 ** 3
plt.plot(my_np1, my_np2, "y*") # grafik çizgi yerine yıldızla gösterilir
```

```python
plt.plot(my_np1, my_np2, "y+")
```

```python
plt.plot(my_np1, my_np2, "y--")
```

```python
plt.plot(my_np1, my_np2, "y+-")
```

```python
plt.subplot(1, 2, 1) # 1 satır, 2 sütun, 1. grafik
plt.plot(my_np1, my_np2, "r--")
plt.subplot(1, 2, 2) # 1 satır, 2 sütun, 2. grafik
plt.plot(my_np1, my_np2, "g*")
```

```python
# figure
my_figure = plt.figure()
my_axes = my_figure.add_axes([1, 1, 0.5, 0.5]) # grafiğin ekranda kapladığı boyutu son iki sayı belirler
my_axes.plot(my_np1, my_np2, "g")
# figure değişkeni üzerinden işlem yaptığımız için bazı fonksiyonlar değişecek
my_axes.set_xlabel("X Data")
my_axes.set_ylabel("Y Data")
my_axes.set_title("X - Y Graph")
```

```python
(my_fig, my_axe) = plt.subplots(2, 2) # ilki satır, ikincisi sütun
```

```python
# tek grafikte iki çizgi
(my_fig, my_axe) = plt.subplots()
my_np3 = np.linspace(0, 15, 20)
my_np4 = my_np3 ** 2
my_axe.plot(my_np4, my_np3, color="#CD621D", alpha=1, linewidth=2) # renk, saydamlık ve kalınlık ayarları
my_axe.plot(my_np4, my_np3 + 2, color="#B81DCD", alpha=0.9, linewidth=3)
my_axe.plot(my_np4, my_np3 + 3, color="green", alpha=0.9, linewidth=0.5, linestyle="-.")
my_axe.plot(my_np4, my_np3 + 4, color="blue", alpha=0.9, linewidth=0.9, linestyle=":")
my_axe.plot(my_np4, my_np3 + 5, color="gray", alpha=0.9, linewidth=0.9, linestyle="--", marker="o")
my_axe.plot(my_np4, my_np3 + 6, color="black", alpha=0.9, linewidth=0.9, linestyle="--", marker="+", markersize=3)
```
